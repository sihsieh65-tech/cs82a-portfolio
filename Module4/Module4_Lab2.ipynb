Import the dataset. I corrected the data in vs code to match the data in the textbook as the csv provided had zeros for the area in rows 7-14. The edited csv file will be uploaded with the rest of the lab. 

```python
import pandas as pd
df = pd.read_csv('allbacks.csv')
df
```

Determining the median cover type by sorting first, then locating the value in the middle of the sorted list. The median cover type was determined to be paperback. This however, has little meaning in context of this analysis. The median value is easily verified as there are only two values and one of them is present exactly one more time than the other leading to a odd number of values. In such a situation the value that is represented one additional time will always be the median value.  

```python
sorted_strings = df['cover'].sort_values().reset_index(drop=True)
mid_index = len(sorted_strings) // 2
median_string = sorted_strings.iloc[mid_index]
print("Median string:", median_string)
```

Checking the median and mean of the book volumes, areas, and weights. Of note, I feel that area and volume are meant to provide similar information with regards to how much space each book occupies by cover type. In our 3-dimensional world, the volume is the more meaningful value and therefore we will prioritize that over the area. As such further analysis will not be done using the area. The overall average weight of books in this library is 706.7 grams which is pretty close to the median weight of 725.0 g. 

```python
df['volume'].median(), df['area'].median(), df['weight'].median()
```

```python
df['volume'].mean(), df['area'].mean(), df['weight'].mean()
```

Calculating the average, median, max and mininmum weights and volumes of the books, grouped by cover type. Based on the grouped means the hardback books weigh notably more than paperback books on average. 

```python
df.groupby('cover').agg(
    average_weight=('weight', 'mean'),
    median_weight=('weight','median'),
    max_weight=('weight', 'max'),
    min_weight=('weight', 'min')
).reset_index()
```

```python
df.groupby('cover').agg(
    average_volume=('volume', 'mean'),
    median_volume=('volume','median'),
    max_volume=('volume', 'max'),
    min_volume=('volume', 'min')
).reset_index()
```

Based on the data I choose to use the median as a basis for comparison as the paperback books show a bit of a skew in the average weight and an even larger skew in the average volume. Based on the median values and assuming I am choosing which cover type to choose based on how to most fully stock my library I would select hardcover books. Since hard coverbooks have a notably smaller median volume I should be able to fit more books than if I chose to stock paperback books.  

However, much of the prompts for this lab focus on the mean weights. If we are basing our decisions on the weights of the average paperback vs the average hardcover I would imagine the goal would be to end up with the lighter average set of books as this would make stocking and moving books around easier. Based on that logic, I would choose to stock paperback books over hardcover as they are lighter on average than the hardcovers. I do note that the average weight of paperback books differs significantly from the median weight and therefore is perhaps not the most reliable measure to use, but it is the one specified in the prompts.
