Below I wrote code for the "while" version of the function. In the first version I limited the loop to run until the number of times it ran which I called, "c" was less than the value of y. So on the first pass c = 0 then since I included c +=1 as part of the function c became 1. Since c < y (1<2) it ran a second pass and added 1 to c again so that c became 2. Then since c < y (2<2) was no longer true, the function terminated. The second version works similarly except that the number of times the loop ran was called "counter" and was originally set to be equal to y. Then on each pass of the function, the counter would have 1 subtracted from it (counter -=1). The loop was set to run until the counter was not equal to 0 (while counter !=0), once it reached zero the function terminated. I also modified these to allow for negative vales for each version. 

```python
def multiply_while(x,y):
    is_negative= (x < 0) ^ (y < 0)
    x,y = abs(x),abs(y)
    result = 0
    c = 0
    
    while c<y:
        result += x
        c += 1
    return -result if is_negative else result

print(multiply_while(58,2))
```

```python
def multiply_while(x,y):
    is_negative= (x < 0) ^ (y < 0)
    x,y = abs(x),abs(y)
    result = 0
    counter = y

    while counter !=0:
        result += x
        counter -= 1
    return -result if is_negative else result

print(multiply_while(58,2))
```
