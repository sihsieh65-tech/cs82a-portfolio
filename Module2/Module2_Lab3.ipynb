Using the categories from the textbook, the following shall be used to label the temperatures:

"Cold" if the temperature is less than 15C;

"Warm" if the temperature is between 15 and 25C (inclusive);

"Hot" if the temperature is greater than 25C

```python
temps= [10,16,23,28,14,22,30]
for t in temps:
    if(t < 15):
        print(t,"Cold")
    elif(t > 25):
        print(t,"Hot")
    else:
        print(t,"Warm")
```

To simplify the logic I used the following code for each respective category:

"Cold" if the temperature is 15C or less: 
    if(t < 15):
        print(t,"Cold")

"Hot" if the temperature is 25C or greater:
    elif(t > 25):
        print(t,"Hot")

"Warm" if the temperature is between 15 and 25C (inclusive):
    else:
        print(t,"Warm")


For "Warm" since the first two conditions are much broader and include all other values outside of the "warm" range, I could simply use the "else" function instead of using a two-part function that checks for higher than or equal to 15 and less than or equal to 25.
