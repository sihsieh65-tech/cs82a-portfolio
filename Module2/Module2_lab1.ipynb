Step 2: Count the warm days

```python
temps = [61, 65, 72, 58, 70, 74, 68]
high = 0
for t in temps:
    if t > 65:
        high = high + 1
print(high)
```

Step 3: Write a temperature converter

```python
def F_to_C(f):
    c = (f-32)*5/9
    return round(c,1)
F_to_C(temps[0])
```
