Create and insert (Appendix C.2)

Following C.2’s example, create a customers table with at least four columns (id, name, city, and one numeric column) using a SQL CREATE TABLE statement, then INSERT at least eight fictional rows. MySQL’s types map to SQLite almost directly: use INTEGER, TEXT, REAL.

```python
import sqlite3
import pandas as pd
conn = sqlite3.connect("customerstable2.db", timeout=5)
cursor = conn.cursor()
cursor.execute("""
    CREATE TABLE IF NOT EXISTS customers (
        id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        city TEXT,
        amount REAL
    )
""")
print("Table created successfully!")
```

```python
conn = sqlite3.connect("customerstable2.db", timeout=5)
cursor = conn.cursor()
multiple_employees = [
    (1, "Andrew Banner", "New York", 950.00),
    (2, "Charlie Danvers", "Chicago", 75.00),
    (3, "Ellie Frank", "Los Angeles", 500.00),
    (4, "Greg Hauser", "Los Angeles", 450.00),
    (5, "Ingrid Jackson", "Los Angeles", 375.00),
    (6, "Kevin Lawson", "Chicago", 100.00),
    (7, "Michael Newsom", "Chicago", 80.00),
    (8, "Otto Pearson", "New York", 1000.00)
]
cursor.executemany(
    "INSERT INTO customers (id, name, city, amount) VALUES (?, ?, ?, ?)", 
    multiple_employees
)
conn.commit()
```
