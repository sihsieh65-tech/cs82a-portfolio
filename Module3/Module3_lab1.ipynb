Step 3: Count the missing values

Counting missing data using .isna().sum(). We identify 12 cases in the price column only.

```python
df.isna().sum()
```

Checking the median price in the column. We will use median and not mean as the mean can skew the data towards any extremes within the data set. Checking the min and max for the price data using .describe() we note a large range between $2.93 and 479.29. Therefore the median seems more appropriate. 

```python
df.describe()
```

Step 4: Impute the missing prices

Filling in missing data with the median price.

```python
df["price"].median()
```

```python
df["price"] = df["price"].fillna(df["price"].median())
```

Checking that the prices were filled. Will likely skip this step (checking in the column only) in favor of re-examining the table.

```python
df["price"]
```

Re-examining the data in the full table.

Step 5: Remove duplicate rows

Now checking for duplicate rows with .duplicated().sum() then dropping duplicates from the table with df.drop_duplicates()

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

Checking that 8 rows were dropped using .shape

Step 6: Repair the zip codes

Correcting the zip codes to re-add leading zeroes. Set the data type to string and setting to give 5 digits, then rechecking the data by displaying it with df. 

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)
```

```python
df
```

Rechecking the data type with .info() to ensure that the zip is now in a str format

```python
df.info()
```

Step 7: Standardize the dates

Changing the dates from strings to date data types and noting a "mixed" format. Then I rechecked the table by displaying with df and using .info() to make sure the datatype is correct. 

```python
df["date"] = pd.to_datetime(df["date"], format="mixed")
```

```python
df
```

```python
df.info()
```

Step 8: Investigate the negative quantities

Identifying the incorrect negative quantities by filtering for df["qty"]<0

```python
df[df["qty"]<0]
```

Negative values are non-sensical in this context. We cannot sell negative units. Therefore I am interpreting this to be a typo and will simply convert the numbers to positive values. 

```python
df["qty"] = df["qty"].abs()
```

Convert negative values to positive ones using the absolute value.

```python
df[df["qty"]<0]
```
